# InsightX – Exploratory Data Analysis Demonstration

This Jupyter Notebook demonstrates the modular programmatic API of **InsightX**. It illustrates loading a dataset, running anomaly and missingness checks, compiling summary statistics, calculating correlations, and performing common clean-up tasks.

### 1. Import local data processing modules

```python
import os
import sys
import pandas as pd

# Ensure the parent directory is in path to import utils
sys.path.append(os.path.abspath('..'))

from utils.helpers import load_dataset, format_size
from utils.preprocessing import detect_issues, fill_missing_value, remove_duplicate_rows
from utils.statistics import calculate_summary_stats, calculate_correlations, get_column_types
```

### 2. Load the seeded Titanic dataset

```python
dataset_path = '../dataset/titanic.csv'
df = load_dataset(dataset_path)
print(f"Loaded Titanic dataset. Shape: {df.shape[0]} rows, {df.shape[1]} columns.")
df.head(5)
```

### 3. Profile Dataset Quality and Detect Issues

```python
issues = detect_issues(df)
print("Detected Dataset Issues Summary:")
print(f"- Duplicate rows count: {issues['duplicate_rows']}")
print("- Missing values per column:", issues['missing_values_by_col'])
print("- Outlier counts by numeric column:", issues['outliers_by_col'])
```

### 4. Descriptive Statistics & Correlations

```python
col_profiles = calculate_summary_stats(df)
print("Age Feature Statistics:")
for stat, val in col_profiles['Age'].items():
    if stat != 'type':
        print(f"  {stat}: {val}")
```

```python
corrs = calculate_correlations(df)
print("Top Positive Correlations:")
for pair in corrs['top_positive']:
    print(f"  {pair['feature1']} & {pair['feature2']} -> Pearson r = {pair['correlation']:.4f}")
```

### 5. Cleaning & Imputation Demo

```python
print(f"Nulls in Age before: {df['Age'].isna().sum()}")
df_imputed = fill_missing_value(df, 'Age', 'median')
print(f"Nulls in Age after median imputation: {df_imputed['Age'].isna().sum()}")
```
